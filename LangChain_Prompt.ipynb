```python
from langchain_openai import ChatOpenAI

# Keep the model configuration in one place so every example uses consistent settings.

def create_chat_model(max_tokens=100):
    return ChatOpenAI(
        model="gpt-4o",             # OpenAI chat model used by these examples.
        temperature=0.5,             # Controls variation in generated responses.
        max_tokens=max_tokens,       # Caps the length of each response.
        top_p=0.2                    # Restricts token selection to likely candidates.
    )
```

## Basic Prompt

A prompt is the input instruction or text sent to a language model. In this first example, the prompt is an unfinished sentence, so the model continues it. `model.invoke(prompt)` sends the text to the model and returns an AI message; `response.content` contains the generated text.

The `max_tokens` setting limits the response length. The value is a maximum, not a guarantee that the model will use every token.

```python
prompt = "The rain is "

model = create_chat_model(12)
response = model.invoke(prompt)

print(f"Prompt: {prompt}\n")
print(f"Response: {response.content}\n")
```

## Zero-Shot Prompting

Zero-shot prompting asks the model to complete a task without showing it any examples. The instruction should state the task and, when useful, the expected labels or output format. Here the model classifies a statement as true or false using its existing knowledge.

This technique is quick to write, but the response may include explanation beyond the requested label. For consistent machine-readable output, explicitly request a concise format.

```python
# Prompt
prompt = """Classify the following statement as true or false: 'Mount Everest is located in Australia'"""

# Generate response
model = create_chat_model(100)
response = model.invoke(prompt)

# Print result
print(f"Prompt: {prompt}")
print(f"Response: {response.content}")
```

## One-Shot Prompting

One-shot prompting includes one input/output example before the new task. The example demonstrates the desired transformation—in this case, translating English into Spanish—and gives the model a pattern to follow.

Use one-shot prompting when a short example can clarify the task or response style. The result can still vary, so the example should be accurate and representative.

```python
# Prompt
prompt = """Here is an example of translating a sentence from English to Spanish:
English: "What time does the train leave?" 
Spanish: "¿A qué hora sale el tren?"

Now, translate the following sentence from English to Spanish:
English: “How is the weather today?”
"""

# Generate response
model = create_chat_model(100)
response = model.invoke(prompt)

# Print result
print(f"Prompt: {prompt}")
print(f"Response: {response.content}")
```

## Few-Shot Prompting

Few-shot prompting provides multiple examples of the task before asking the model to handle a new input. Here, the examples associate statements with emotion labels, which demonstrates the classification categories and the expected mapping.

Examples can improve consistency when the task or labels are unfamiliar. Keep them clear, balanced, and relevant; examples also use context tokens, leaving less room for the answer.

```python
# Prompt
prompt = """Here are few examples of classifying emotions in statements:
"I finally got the job offer!" → Joy "The app crashed right before I could save my work." → Frustration "I really miss my old school friends." → Sadness

Now, classify the following statement as Joy, Frustration, or Sadness:
Statement: "I won the lottery!"
"""

# Generate response
model = create_chat_model(100)
response = model.invoke(prompt)

# Print result
print(f"Prompt: {prompt}")
print(f"Response: {response.content}")
```

## Step-by-Step Reasoning Prompt

This prompt asks the model to break a multi-step arithmetic problem into steps before giving a result. It illustrates how explicitly requesting intermediate work can encourage a structured explanation.

A generated explanation is not proof that the answer is correct. For important calculations, independently verify the steps or use code/tools for deterministic arithmetic.

```python
# Prompt
prompt = """Consider the problem: 'A bakery had 40 cupcakes. It sold 28 in the morning and baked 15 more in the afternoon. 
How many cupcakes does it have now?'
Break down each step of your calculation
"""

# Generate response
model = create_chat_model(500)
response = model.invoke(prompt)

# Print result
print(f"Prompt: {prompt}")
print(f"Response: {response.content}")
```

## Self-Consistency Prompting

Self-consistency prompting asks for several independent solution paths and then a comparison of their results. The example uses an age problem: solving it through different approaches can help expose inconsistencies in a single line of reasoning.

This notebook requests multiple explanations in one response; it does not automatically run separate samples or perform a majority vote. For a true self-consistency workflow, generate multiple independent answers and compare their final results.

```python
# Prompt
prompt = """When I was 8, my brother was a quarter of my age. Now I am 40. How old is my brother?
Provide three independent calculations and explanations, then determine the most consistent result.
"""

# Generate response
model = create_chat_model(500)
response = model.invoke(prompt)

# Print result
print(f"Prompt: {prompt}")
print(f"Response: {response.content}")
```

## Prompt vs. Prompt Template

A plain prompt is a completed string sent to a model. A `PromptTemplate` is a reusable pattern with named placeholders, such as `{style}` and `{product}`. Calling `format` supplies values for those placeholders and produces the final prompt.

The last example connects prompt formatting, model invocation, and output parsing into a LangChain runnable sequence. `RunnableLambda` adapts the formatting function, `ChatOpenAI` generates a response, and `StrOutputParser` extracts the response text. The dictionary passed to `invoke` must provide values for every required template variable.

```python
from langchain_core.prompts import PromptTemplate, ChatPromptTemplate
from langchain_core.runnables import RunnableLambda
from langchain_core.output_parsers import StrOutputParser

template = """Write a {style} product description for {product}."""
prompt = PromptTemplate.from_template(template)
prompt 
```

```python
prompt.format(style="Catchy", product="T-shirt")
```

```python
def format_prompt(variables):
    return prompt.format(**variables)
```

```python
# Create the chain with explicit formatting
product_chain = (
    RunnableLambda(format_prompt)
    | create_chat_model(100) 
    | StrOutputParser()
)

response = product_chain.invoke({"style": "Catchy", "product": "T-shirt"})
print(response)
```
